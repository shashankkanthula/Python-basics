```python
#example for matplotlib
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([0,5])
ypoints= np.array([0,250])
plt.plot(xpoints,ypoints,)
plt.show()
```

```python
#for marking the points of the line
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([1,2,6,8])
ypoints= np.array([3,8,1,10])
plt.plot(xpoints,ypoints,'o')
plt.show()
```

```python
#markers (combination of line and the points)
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([1,2,6,8])
ypoints= np.array([3,8,0,10])
plt.plot(xpoints,ypoints,marker='o')
plt.show()
```

```python
#linestyle (dashed,dotted)
import matplotlib.pyplot as plt
import numpy as np
ypoints= np.array([3,8,1,10])
plt.plot(ypoints,linestyle='dotted') #if we change the linestyle the graph is going to change
plt.show()
```

```python
#labels and title
import matplotlib.pyplot as plt
import numpy as np
ypoints= np.array([3,8,1,10])
plt.plot(ypoints)
plt.title('sales')
plt.xlabel('years')
plt.ylabel('points')
plt.show()
```
