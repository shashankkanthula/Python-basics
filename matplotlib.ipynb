```python
#example for matplotlib
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([0,5])
ypoints= np.array([0,250])
plt.plot(xpoints,ypoints,)
plt.show()
```

```python
#for marking the points of the line
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([1,2,6,8])
ypoints= np.array([3,8,1,10])
plt.plot(xpoints,ypoints,'o')
plt.show()
```

```python
#markers (combination of line and the points)
import matplotlib.pyplot as plt
import numpy as np
xpoints= np.array([1,2,6,8])
ypoints= np.array([3,8,0,10])
plt.plot(xpoints,ypoints,marker='o')
plt.show()
```

```python
#linestyle (dashed,dotted)
import matplotlib.pyplot as plt
import numpy as np
ypoints= np.array([3,8,1,10])
plt.plot(ypoints,linestyle='dotted') #if we change the linestyle the graph is going to change
plt.show()
```

```python
#labels and title
import matplotlib.pyplot as plt
import numpy as np
ypoints= np.array([3,8,1,10])
plt.plot(ypoints)
plt.title('sales')
plt.xlabel('years')
plt.ylabel('points')
plt.show()
```

```python
#grid (adding grid to the graph)
import matplotlib.pyplot as plt
import numpy as np
ypoints= np.array([3,8,1,10])
plt.plot(ypoints)
plt.title('sales')
plt.xlabel('years')
plt.ylabel('points')
plt.grid() #in the place of show we are going to replace it with grid
```

```python
#sub plot
import matplotlib.pyplot as plt
import numpy as np
y1= np.array([3,8,1,10])
y2= np.array([6,2,7,11])
plt.subplot(1,2,1)
plt.plot(y1)
plt.title('sales')
x1= np.array([0,1,2,3])
x2= np.array([0,1,2,3])
plt.subplot(1,2,2)
plt.plot(x1,x2)
plt.plot()
plt.title('income')

#suptitle is the main heading for the both graphs
plt.suptitle('sales and income')
plt.show()
```

```python
#Scatter plots (used to represent with a dot for each observation)
import matplotlib.pyplot as plt
import numpy as np
x= np.array([5,7,8,7,2,1])
y= np.array([99,86,87,88,111,86])
plt.scatter(x,y)
plt.show()
```

```python
#Bars
import matplotlib.pyplot as plt
import numpy as np
x= np.array(['A','B','C','D'])
y= np.array([3,8,1,10])
plt.xlabel('bar number')
plt.ylabel('points')
plt.bar(x,y)  #bar is for vertical and the (barh) is for horizontal
plt.show()

plt.barh(x,y)
plt.show()
```

```python
#Histogram (freq distribution)
import matplotlib.pyplot as plt
import numpy as np
x= np.random.normal(170,10,250)
plt.hist(x)
plt.xlabel('height')
plt.ylabel('frequency')
plt.title('Histogram')
plt.show()
```

```python
#pie chart
import matplotlib.pyplot as plt
import numpy as np
y= np.array([35,25,25,15])
mylabels= ['Apples','Bananas','Cherries','Dates']
plt.pie(y,labels=mylabels)
plt.show()
```
